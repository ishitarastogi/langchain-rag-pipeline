```python
import pymupdf

doc = pymupdf.open("../data/text_files/Reliance.pdf")

print(doc.page_count)
```

```python
for page_num in range(35):
    text = doc[page_num].get_text()
    head = " ".join(text[:100].split())
    print(f"{page_num:3d} | {len(text):5d} | {head}")
```

```python
counts = {}

for page_num in range(27, doc.page_count):
    for line in doc[page_num].get_text().split("\n"):
        line = line.strip()
        if 3 < len(line) < 60:
            counts[line] = counts.get(line, 0) + 1

top = sorted(counts, key=lambda l: counts[l], reverse=True)
for line in top[:20]:
    print(f"{counts[line]:4d}x  {line[:60]}")
```

```python
for page_num in range(24, 32):
    text = doc[page_num].get_text()
    head = " ".join(text[:150].split())
    print(f"{page_num:3d} | {len(text):5d} | {head}")
```

```python
page_lengths = {}

for page_num in range(doc.page_count):
    text = doc[page_num].get_text()
    page_lengths[page_num] = len(text)

for page_num, length in page_lengths.items():
    print(f"Page {page_num}: {length}")
```

```python
sorted_pages = sorted(page_lengths, key=lambda p: page_lengths[p])
print(sorted_pages[:10])
print("SHORTEST:")
for p in sorted_pages[:10]:
    print(f"  Page {p}: {page_lengths[p]} chars")

print("LONGEST:")
for p in sorted_pages[-10:]:
    print(f"  Page {p}: {page_lengths[p]} chars")
```
