```python
from docling.document_converter import DocumentConverter
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.datamodel.base_models import InputFormat
from docling.document_converter import PdfFormatOption

# Faster settings
pipeline_options = PdfPipelineOptions()
pipeline_options.do_ocr = False              # Very important - disables OCR
pipeline_options.do_table_structure = True   # Keep table detection

converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)
    }
)

print("Starting conversion...")
result = converter.convert("../data/text_files/Reliance.pdf")
doc = result.document

print(f"Done! Total tables found: {len(doc.tables)}")
```

```python
for i, table in enumerate(doc.tables[:10]):  # Start with only first 10 tables
    page_num = table.prov[0].page_no if table.prov else "Unknown"
    print(f"\n=== Table {i+1} (Page {page_num}) ===")
    print(table.export_to_markdown(doc=doc))
```
